1. Describe different types of data sources used in ETL with suitable examples.
- n ETL (Extract, Transform, Load), data can be extracted from multiple types of sources:

1. Relational Databases

Structured data stored in tables.

Examples: MySQL, Oracle Database, Microsoft SQL Server

Used in banking systems, ERP systems, CRM applications.

2. Flat Files

Simple text-based files.

CSV, TXT files

Example: Sales_data.csv containing transaction records.

3. Spreadsheets

Used for small to medium data storage.

Microsoft Excel (.xlsx)

4. APIs (Web Services)

Used to fetch live or real-time data.

Example: Twitter API for tweets, Google Maps API for location data.

5. Cloud Storage

Data stored in cloud platforms.

Amazon S3, Google Cloud Storage

6. NoSQL Databases

Unstructured or semi-structured data.

MongoDB, Cassandra

2. What is data extraction? Explain its role in the ETL pipeline.
-  Data Extraction is the process of collecting data from various source systems for further processing.

Role in ETL:

First step of ETL.

Ensures accurate data collection.

Identifies required data fields.

Maintains data consistency.

Prepares raw data for transformation.

Without proper extraction, the entire ETL process fails because incorrect or incomplete data leads to poor analysis results.

-  Feature           CSV                    Excel

 Size               Lightweight             Larger

4. Explain the steps involved in extracting data from a relational database.
- Identify source database.

Establish connection (using credentials).

Write SQL query (SELECT statement).

Filter required records.

Export data (CSV, staging table, etc.).

Validate extracted data.

Log extraction process.

Example:
SELECT customer_id, name, city
FROM customers
WHERE status = 'active';


5. Explain three common challenges faced during data extraction.
- 1. Data Quality Issues

Missing values, duplicates, incorrect formats.

2. Large Data Volume

Extraction of millions of records may slow performance.

3. Data Security & Access

Restricted permissions or encrypted databases.

Other challenges:

Network failures

Schema changes

Inconsistent data formats

6. What are APIs? Explain how APIs help in real-time data extraction.
- API (Application Programming Interface) allows communication between software systems.

Role in Real-Time Extraction:

Fetch live data.

Automate data retrieval.

Support JSON/XML format.

Enable incremental updates.

Example:

Stripe API for payment data.

Weather API for live weather updates.

APIs are essential in modern ETL pipelines for streaming and real-time analytics.

7. Why are databases preferred for enterprise-level data extraction?
- Structured and reliable data.

Support for SQL queries.

Better security controls.

Transaction management (ACID properties).

High performance and scalability.

Backup and recovery features.

Enterprise tools like SAP and Salesforce store large-scale business data in databases.

8. What steps should an ETL developer take when extracting data from large CSV files (1GB+)?
- Use chunk processing (read in batches).

Avoid loading entire file into memory.

Use efficient libraries (e.g., pandas with chunksize).

Validate data schema before loading.

Remove unwanted columns early.

Use compression (gzip).

Monitor memory and CPU usage.

Use parallel processing if required.
